```python
import pandas as pd
import numpy as np
```

### 1. Series: 
###    A one diamension array which can hold any data type.
    
### Syntax:
###    pd.Series(data , Index)

```python
ser  = pd.Series([2,4,5,6,7])
```

```python
print(ser)
```

```python
ser1 = pd.Series([1,2,3,4,5] , index = ['a','b','c','d','e'])
print(ser1)
```

```python
a = (23,45,76,22,53)
b = ['a','b','c','d','e']

ser2 = pd.Series(a , index = b)
print(ser2)
```

### DataFrame : 

### It is a two diamension array which contains rows and columns.

### syntax:
###      pd.DataFrame(data , index)

```python
data = {
    'Name':['Ved','Eve'],
    'Age' :[25 , 23]
}

df1 = pd.DataFrame(data)
print(df1)
```

```python
Data = {
    'ID' : [1,2,3,4,5],
    'Name' : ['Ved','Eve',"Alena","David",'Stephen'],
    'Age' : [25,np.nan,24,26,28],
    'Gender' : ['M','F','F','M','M'],
    'Salary' : [40000,50000,35000,np.nan,45000]
}

df = pd.DataFrame(Data)
```
