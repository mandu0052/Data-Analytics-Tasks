```python
# indexing ---> TO access the specific element of array

arr = np.array([54,6,43,32,64])
print(arr[1])
```

```python
arr2 = np.array([[1,2,3,4],[5,6,7,8]])
print(arr2[1,3])
```

```python
#slicing
print(arr[0:3])
```

```python
print(arr2[0,1:3])
```

```python
# Copy and view 

x = arr.copy()
arr[0] = 5
print(arr)
print(x)
```

```python
y = arr.view()
arr[1] = 56
print(arr)
print(y)
```
